## 1. Setup - run before recording
Run once in a connected runtime; repeat safely if necessary (a new temporary folder is created).


```python
import os, subprocess, sys, tempfile
from pathlib import Path

work = Path(tempfile.mkdtemp(prefix="nvda-video2-"))
repo = work / "repo"
subprocess.run([
    "git", "clone", "--quiet", "--depth", "1", "--branch",
    "project-1-nvidia-recording-charts-2026-10-19",
    "https://github.com/JP161632/FIN43900-Fall2026.git", str(repo)
], check=True, capture_output=True, text=True)
commit = subprocess.check_output(["git", "-C", str(repo), "rev-parse", "HEAD"], text=True).strip()
assert commit == "6d0d76de08f034e02dfc0a736bcab8ab421f7277", "Unexpected source version"
project = repo / "projects" / "project-1"
env = work / "clean-env"
subprocess.run([sys.executable, "-m", "venv", "--without-pip", str(env)], check=True)
clean_python = env / ("Scripts/python.exe" if os.name == "nt" else "bin/python")
output = work / "video2-output"
print("Ready. Frozen NVIDIA project loaded; new no-pip environment created.")
print("Run cell 2 during your recording.")
```

## 2. Actual project run - record this cell
Expected success: PASS15/15, base$118.58, WATCH/DEFER. Record actual output rather than assuming success.


```python
from datetime import datetime, timezone
import json, platform

started = datetime.now(timezone.utc).isoformat()
command = [str(clean_python), str(project / "run.py"), "--all", "--output", str(output)]
result = subprocess.run(command, capture_output=True, text=True)
print(result.stdout, end="")
if result.stderr:
    print(result.stderr, end="")
output.mkdir(parents=True, exist_ok=True)
(output / "recording-run-log.json").write_text(json.dumps({
    "started_utc": started, "source_commit": commit,
    "runtime": platform.platform(), "python": sys.version,
    "isolation": "new no-pip virtual environment; fresh copy of frozen project",
    "exit_code": result.returncode, "stdout": result.stdout, "stderr": result.stderr,
    "scope": "Actual execution log; does not establish prediction authorship or video existence"
}, indent=2) + chr(10), encoding="utf-8")
result.check_returncode()
```
